### Task 1 - Weekend Activities as Sets

```python
actSat=set()
active=True
while active:
    user_activity=input('Enter an activity from last Saturday. Type "done" to stop.')
    if user_activity.lower()=="done":
        active=False
        break
    else:
        actSat.add(user_activity)

actSun=set()
active=True
while active:
    user_activity=input('Enter an activity from last Sunday. Type "done" to stop.')
    if user_activity.lower()=="done":
        active=False
        break
    else:
        actSun.add(user_activity)

print(actSat)
print(actSun)
print(actSat.intersection(actSun))
print(actSat.union(actSun))
```

### Task 2 - Triangle Calculations

```python
def triangle_solve(leg1, leg2):
    leg3=(leg1**2+leg2**2)**(1/2)
    perimeter=leg1+leg2+leg3
    area=leg1*leg2/2
    answer=(leg3, perimeter, area)
    return answer
triangle_solve(3,4)
```

### Task 3 - Calculator Class

```python
class Calculator():
    def __init__(self):
        self.previous_result=0
        self.total_calculations=0
    def add(self,a,b):
        self.a=a
        self.b=b
        self.previous_result=self.a+self.b
        self.total_calculations+=1
        return self.a+self.b
    def subtract(self,a,b):
        self.a=a
        self.b=b
        self.previous_result=self.a-self.b
        self.total_calculations+=1
        return self.a-self.b
    def multiply(self,a,b):
        self.a=a
        self.b=b
        self.previous_result=self.a*self.b
        self.total_calculations+=1
        return self.a*self.b
    def divide(self,a,b):
        self.a=a
        self.b=b
        self.previous_result=self.a/self.b
        self.total_calculations+=1
        return self.a/self.b
    def power(self,a,b):
        self.a=a
        self.b=b
        self.previous_result=self.a**self.b
        self.total_calculations+=1
        return self.a**self.b
    def info(self):
        print(f'Previous result = {self.previous_result} and total calculations = {self.total_calculations}.')
calc1=Calculator()
calc1.add(1,2)
calc1.add(8,2)
calc1.subtract(10,3)
calc1.multiply(8,6)
calc1.divide(8,2)
calc1.power(8,3)
calc1.info()
```
