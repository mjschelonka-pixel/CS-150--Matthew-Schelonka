## Task 1: Input Positive Integer and Factors

```python
num1=int(input('Please enter a positive integer:'))
primecounter=0
for num in range(1,num1+1):
    if num1%num==0:
        print(f'{num} is a factor of {num1}')
        primecounter+=1
if primecounter==2:
    print('Your number is prime')
else:
    print('Your number is not prime')
```

## Task 2: Language Translator

```python
english_dict={'hello':'hola','name':'nombre','flower':'flor','ball':'pelota','friend':'amigo','place':'lugar','school':'escuela','number':'numero','phone':'telefono','glasses':'gafas'}
active=True
while active:
    user_word=str(input('Enter a word to translate from English to Spanish (hit enter to stop):'))
    if user_word in english_dict:
        print(f'The Spanish word for {user_word} is {english_dict[user_word]}')
    elif user_word=='':
        active=False
    elif user_word not in english_dict:
        print(f'The Spanish word for {user_word} is not in this dictionary')
```

## Task 4: Inventory Dictionary and Prices

```python
inventory={'hammer':[15,12.49],'wrench':[20,7.99],'screwdriver':[40,3.99],'nailgun':[10,36.59],'padlock':[25,5.99],'roll_of_duct_tape':[30,4.95]}
while True:
  print("\n--- Inventory Menu ---")
  print("1. Check item availability and price")
  print("2. Find items below a budget")
  print("3. Total quantity of items across inventory")
  print("4. Total price of all items in inventory")
  print("5. Quit")
  choice = input("Please select an option (1-5): ")
  if choice == "1":
    purchase = input("Enter an item name: ")
    if purchase not in inventory:
      print("That item is not available.")
    else:
      available = inventory[purchase][0]
      price = inventory[purchase][1]
      print(f"The amount available of {purchase} is {available} and is priced"f" at ${price}.")
  elif choice == "2":
    budget = float(input("Enter in your budget: "))
    print(f"Items with an individual cost below ${budget}:")
    found = False
    for item, details in inventory.items():
      price = details[1]
      if price < budget:
        print(f"- {item} (${price})")
        found = True
    if not found:
      print("No items found below this budget.")
  elif choice == "3":
    total_qty = sum(details[0] for details in inventory.values())
    print(f"The total quantity of items available across all names is: {total_qty}")
  elif choice == "4":
    total_price = sum(details[0] * details[1] for details in inventory.values())
    print(f"The total price of all items across the inventory is:"f" ${total_price:.2f}")
  elif choice == "5":
    print("Goodbye!")
    break
  else:
    print("Invalid selection. Please choose a number from 1 to 5.")
```
