## Task 1: Types of Triangles

```python
side1=int(input('Please enter the length of the 1st side of a triangle:'))
side2=int(input('Please enter the length of the 2nd side of the triangle:'))
side3=int(input('Please enter the length of the 3rd side of the triangle:'))
if side1==side2 and side1==side3:
    print('You entered in an equilateral triangle')
elif side1==side2 or side1==side3 or side2==side3:
    print('You entered in an isosceles triangle')
else:
    print('You entered in a scalene triangle')
```

## Task 2: Student Grades and At-Risk Status

```python
student_grades=[]
i=0
counter=0
while counter<5:
    grade=float(input("Enter in a grade"))
    student_grades.append(grade)
    if student_grades[counter]<65.0:
        i+=1
    counter+=1
print('The highest grade entere was:',max(student_grades))
print('The lowest grade entered was:',min(student_grades))
print('The average grade was:',sum(student_grades)/len(student_grades))
print('Students at risk in this course:',i)
```

## Task 3: Ranges and Multiples

```python
mult_of_2=[]
mult_of_3=[]
for num in range(1, 100):
    if num%2==0:
        mult_of_2.append(num)
print(mult_of_2,'\n')
for num in range(1, 100):
    if num%3==0:
        mult_of_3.append(num)
print(mult_of_3,'\n')
mult_of_both=mult_of_2+mult_of_3
print(mult_of_both,'\n')
mult_of_either=[]
for num in range(1,100):
    if num%2==0:
        mult_of_either.append(num)
        continue
    if num%3==0:
        mult_of_either.append(num)
        continue
print(mult_of_either)
```

## Task 4: Shopping List

```python
shopping_list=[]
i=0
active=True
while active:
    item=input('Enter an item you need from the grocery store, enter done if nothing else is needed:')
    if item!='done':
        shopping_list.append(item)
        i+=1
    else:
        active=False
    if i==7:
        active=False
print('Total number of items on list:',i)
shopping_list_rev=list(reversed(shopping_list))
print(shopping_list_rev)
print(sorted(shopping_list))
for i in range(len(shopping_list)):
    for j in range(i+1, len(shopping_list)):
        if shopping_list[i]==shopping_list[j]:
            print('Duplicate found:',shopping_list[i])
        else:
            print('No duplicates found')
```
