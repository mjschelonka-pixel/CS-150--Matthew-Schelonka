```python
import random

# Number Of Wins And Losses (Set ONCE, Before Any Game Starts, So The Record Is Kept Across Games)
wins=0
losses=0 

print('Welcome To Mastermind! You Are Supposed To Figure Out The Secret 4 Digit Code.  Each Guess You Will Enter 4 Numbers, Starting From Left To Right')

# Outer Loop: One Pass= One Full Game.  The Game Keeps Going Until The User Says They Don't Want To Play Again.  
playing=True
while playing: 
    # Creates A 4 Digit Secret Code Using The Randrange Function To Assign NonDuplicate Numbers To a List Called Secret_Code.  
    Secret_Code=[]
    while len(Secret_Code)<4: 
        code=random.randrange(0,10)
        if code not in Secret_Code: 
            Secret_Code.append(code)
    # Number Of Guesses The User Has
    attempts=10

    # Starts As An Empty List So The Win Check Works Even If The First Guess Is Invalid
    guess_code=[]

    # Inner Loop: Keeps The Game Running As Long As The Guess Is Not Correct And The User Has Attempts Left
    while attempts>0 and guess_code!=Secret_Code: 
        #Take User Input To Get 4 Integer Numbers For Their Guess
        guess=input('Enter Your 4 Digit Integer Number Guess:').strip()
        if len(guess)!=4: 
            print('Your Guess Must Be Four Digits Long!')
            continue
        try: 
            # Take The 4 Number Guess And Create It Into A List To Compare To Secret_Code
            guess_code=[int(digit) for digit in guess]
        except ValueError: 
            print('Your Guess Must Be An Integer!')
            continue
        # Number Of Correct Numbers In Position, And Correct Numbers Not In Correct Positions
        correctposition=0
        correctnum=0

        # For Loop That Checks The Guess_Code Variable To The Secret_Code Variable To See If They Match
        for i in range (4):  
            if guess_code[i]==Secret_Code[i]: 
                correctposition+=1
        # Count Each Digit Only Once, So A Guess Like 1111 Doesn't Score 1 In Every Slot
        for digits in set (guess_code): 
            if digits in Secret_Code: 
                correctnum+=1
        correctnum-=correctposition

        print(f'Correct Numbers In Position: {correctposition} And Correct Numbers In Wrong Position: {correctnum}')
        attempts-=1
        print(f'Attempts Left: {attempts}')

    # If Statements That Check If The User Has Won Or Lost The Game
    if guess_code==Secret_Code: 
        wins+=1
        print(f'You Have Won! Your Total Score Is Now {wins} Win(s) And {losses} Loss(es).')
    else: 
        losses+=1
        Secret=''.join(str(d) for d in Secret_Code)
        print(f'You Have Lost. The Code Was {Secret}. Your Total Score Is Now {wins} Win(s) And {losses} Loss(es).')

    # Ask After Every Game (Win Or Lose) Whether To Play Again
    play=input('Would You Like To Play Again? Yes Or No:')
    if play.strip().lower()!='yes': 
        print('Thank You For Playing! Have A Lovely Day!')
        playing=False
        
        
        
        
            
        
    
    
   
```
